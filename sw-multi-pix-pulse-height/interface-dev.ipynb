```python
%load_ext autoreload

from pathlib import Path
import numpy as np

%autoreload 2
from  panoseti_interface import PanosetiRun, PFFSequence, filter_pff_sequence
```

```python
test_out_dir = Path("out")
data_dir = Path("palomar_data")
run_dir = "obs_Palomar.start_2026-01-20T02:24:17Z.runtype_obs-test.pffd"
```

```python
# 1. Load the Run
run_path = "palomar_data/obs_Palomar.start_2026-01-20T02:24:17Z.runtype_obs-test.pffd"
try:
    run = PanosetiRun(run_path)
except FileNotFoundError:
    print("Run directory not found, check path.")
    exit(1)

print("Available Products:", run.list_products())
# Expected: ['dp_ph1024.bpp_2.module_253', 'dp_ph1024.bpp_2.module_252', ...]

```

```python
# 2. Select a PH1024 stream
target_product = "dp_ph1024.bpp_2.module_250"
if target_product in run.products:
    seq = run.get_product(target_product)

    # 3. Preview Data
    print(f"Total frames: {len(seq)}")
    if len(seq) > 0:
        seq.preview_frame(1)

    # 4. Define a Filter Function
    # Example: Keep frames where the max pulse height is > 1000 (Simple trigger simulation)
    def high_energy_filter(header: dict, img: np.ndarray) -> bool:
        # Check if image has significant signal
        # Pulse height data is signed int16 (baseline subtracted)
        return np.max(img) > 500

    # 5. Run Filter
    out_file = "TEST-filtered_output.pff"
    filter_pff_sequence(seq, out_file, high_energy_filter)

    # Verify Output
    filtered_seq = PFFSequence([out_file])
    print(f"Filtered file has {len(filtered_seq)} frames.")
    if len(filtered_seq) > 0:
        filtered_seq.preview_frame(0)
```
