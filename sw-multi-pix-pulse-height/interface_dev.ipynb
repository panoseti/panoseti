```python
# Assume 'run_obj' is an instance of PanosetiRun initialized earlier

def create_visualizer(run_obj):
    # 1. Selectors
    products = run_obj.list_products()
    dropdown = widgets.Dropdown(options=products, description='Stream:')

    # 2. Controls
    slider = widgets.IntSlider(min=0, max=0, step=1, description='Frame:', continuous_update=False)
    btn_prev = widgets.Button(description="< Prev", layout=widgets.Layout(width='80px'))
    btn_next = widgets.Button(description="Next >", layout=widgets.Layout(width='80px'))

    # 3. Output
    out = widgets.Output()

    # State
    current_seq = None

    def update_product(change):
        nonlocal current_seq
        prod_name = change['new']
        current_seq = run_obj.get_product(prod_name)
        slider.max = len(current_seq) - 1
        slider.value = 0
        update_plot(0)

    def update_plot(idx):
        if current_seq is None: return

        header, img = current_seq.get_frame(idx)

        out.clear_output(wait=True)
        with out:
            # Efficient Matplotlib
            fig, ax = plt.subplots(figsize=(6, 5))

            # Robust mapping for Module vs Quabo
            if img.shape == (32, 32):
                # Draw lines to separate quabos
                ax.axvline(16, color='w', linestyle='--', alpha=0.5)
                ax.axhline(16, color='w', linestyle='--', alpha=0.5)
                title_prefix = "Module"
            else:
                title_prefix = "Quabo"

            # Determine correct IDs from filename metadata or header
            mod_id = current_seq.meta.get('module_id', '?')

            # Display
            im = ax.imshow(img, cmap='inferno', origin='upper')
            plt.colorbar(im, ax=ax, label='ADC Counts')

            # Extract timestamp from Pydantic header
            ts = getattr(header, 'timestamp', 'N/A')
            if isinstance(ts, float): ts = f"{ts:.3f}"

            ax.set_title(f"{title_prefix} {mod_id} | Frame {idx} | T={ts}\nProduct: {dropdown.value}")
            plt.show()

    def on_slider(change):
        update_plot(change['new'])

    def on_next(b):
        if slider.value < slider.max: slider.value += 1

    def on_prev(b):
        if slider.value > 0: slider.value -= 1

    # Wire events
    dropdown.observe(update_product, names='value')
    slider.observe(on_slider, names='value')
    btn_next.on_click(on_next)
    btn_prev.on_click(on_prev)

    # Layout
    controls = widgets.HBox([btn_prev, slider, btn_next])
    ui = widgets.VBox([dropdown, controls, out])

    # Initialize
    if products:
        update_product({'new': products[0]})

    display(ui)
```

```python
# Usage in Notebook:
# run_obj = PanosetiRun("path/to/data")
create_visualizer(run)
```

# The Geometric Algorithm

We define the filtering logic mathematically. This algorithm selects images containing compact, bright signal clusters while rejecting diffuse noise.

Definitions:
* Let $I_{i,j}$ be the calibrated intensity (PE) at pixel $(i,j)$.
* Let $T$ be the detection threshold.
* Let $(x_{grid}, y_{grid})$ be the coordinate matrices of the detector.

### Step 1: Thresholding & Weighting
We define a binary mask $M$ and an intensity weight matrix $W$:$$M_{i,j} = \begin{cases} 1 & \text{if } I_{i,j} \ge T \\ 0 & \text{otherwise} \end{cases}$$$$W_{i,j} = I_{i,j} \cdot M_{i,j}$$

### Step 2: Intensity-Weighted Center of Mass (CoM)
Unlike a geometric centroid, this CoM is pulled towards the brightest pixels:$$X_{com} = \frac{\sum_{i,j} j \cdot W_{i,j}}{\sum_{i,j} W_{i,j}}, \quad Y_{com} = \frac{\sum_{i,j} i \cdot W_{i,j}}{\sum_{i,j} W_{i,j}}$$

### Step 3: Compactness Check

We calculate the Euclidean distance $D_{i,j}$ of every pixel from the CoM:$$D_{i,j}^2 = (j - X_{com})^2 + (i - Y_{com})^2$$We count how many triggered pixels fall within radius $R$:$$K_{compact} = \sum_{i,j} M_{i,j} \cdot \mathbb{I}(D_{i,j} < R)$$(Where $\mathbb{I}$ is the indicator function)

### Step 4: Acceptance Condition
The frame is saved if:$$(\sum_{i,j} M_{i,j} \ge N_{min}) \quad \text{AND} \quad (K_{compact} \ge K_{min})$$

```python
import ipywidgets as widgets
import matplotlib.pyplot as plt
import matplotlib.patches as patches
import numpy as np
from IPython.display import display

from multi_pix_filter import JAXFilterKernel

# Enable LaTeX rendering
plt.rcParams['text.usetex'] = True
# Set the font family to match LaTeX's default Computer Modern font
plt.rcParams['font.family'] = 'serif'
plt.rcParams['font.serif'] = ['Computer Modern Roman']

def create_mahalanobis_visualizer(run_object):
    """
    Visualizes Raw Data -> Calibrated PE -> Mahalanobis Ellipse Fitting.
    """
    valid_prods = [p for p in run_object.list_products() if 'img' in p or 'ph' in p]
    drop_stream = widgets.Dropdown(options=valid_prods, description='Stream:')
    
    # Simulating Calibration Params
    # We add 700 baseline and divide by 60 gain as requested
    s_base   = widgets.FloatSlider(value=700.0, min=0, max=1000, description='Sim Base')
    s_gain   = widgets.FloatSlider(value=60.0, min=1, max=100, description='Sim Gain')
    
    s_thresh = widgets.FloatSlider(value=5.0, min=0.5, max=50, step=0.5, description='PE Thresh')
    s_sigma  = widgets.FloatSlider(value=2.5, min=1.0, max=5.0, step=0.1, description='Sigma')
    s_frame  = widgets.IntSlider(value=0, description='Frame')
    
    btn_next = widgets.Button(description="Next >", layout=widgets.Layout(width='80px'))
    btn_prev = widgets.Button(description="< Prev", layout=widgets.Layout(width='80px'))
    
    out = widgets.Output()
    
    # State
    current_seq = None
    kernel = None
    
    def init_stream(name):
        nonlocal current_seq, kernel
        current_seq = run_object.get_product(name)
        s_frame.max = len(current_seq) - 1
        s_frame.value = 0
        kernel = JAXFilterKernel(shape=current_seq.frame_config.image_shape)
        update_view()

    def update_view(_=None):
        if not current_seq: return
        idx = s_frame.value
        
        # 1. Fetch Raw Data
        raw_batch = current_seq.get_image_array(idx, 1)
        if len(raw_batch) == 0: return
        raw_img = raw_batch[0].astype(np.float32)
        
        # 2. Setup Calibration Maps (Simulated as scalar or arrays)
        # Using scalars here for simplicity, but kernel supports full (16,16) arrays
        base_map = np.full(raw_img.shape, s_base.value)
        gain_map = np.full(raw_img.shape, s_gain.value)
        
        # 3. Run JAX Kernel
        pe_img, mask, debug = kernel.run(
            raw_img, base_map, gain_map, 
            threshold=s_thresh.value, sigma=s_sigma.value
        )
        
        # Unpack Debug
        com_x, com_y = debug[0], debug[1]
        cov_xx, cov_yy, cov_xy = debug[2], debug[3], debug[4]
        intensity_sum = debug[5]
        
        # 4. Calculate Ellipse Geometry for Plotting
        # Eigenvalues of Covariance Matrix
        term = np.sqrt(((cov_xx - cov_yy)**2) + 4 * cov_xy**2)
        lambda1 = (cov_xx + cov_yy + term) / 2
        lambda2 = (cov_xx + cov_yy - term) / 2
        
        # Radii (scaled by sigma)
        # Standard deviation is sqrt(variance/eigenvalue)
        r_major = np.sqrt(np.maximum(0, lambda1)) * s_sigma.value * 2 
        r_minor = np.sqrt(np.maximum(0, lambda2)) * s_sigma.value * 2
        
        # Angle
        theta = 0.5 * np.arctan2(2*cov_xy, cov_xx - cov_yy)
        theta_deg = np.degrees(theta)

        # 5. Plotting
        out.clear_output(wait=True)
        with out:
            fig = plt.figure(figsize=(10, 5), constrained_layout=True)
            gs = fig.add_gridspec(1, 2, width_ratios=[1.2, 1])
            ax = fig.add_subplot(gs[0])
            
            # Show Calibrated PE Image
            im = ax.imshow(pe_img, cmap='inferno', origin='upper')
            cb = plt.colorbar(im, ax=ax, label="Photoelectrons (PE)")
            
            # Lock Axes
            H, W = pe_img.shape
            ax.set_xlim(-0.5, W-0.5)
            ax.set_ylim(H-0.5, -0.5)
            
            # Overlay: Triggered Pixels
            ty, tx = np.where(pe_img >= s_thresh.value)
            ax.scatter(tx, ty, c='cyan', marker='x', alpha=0.7, label='Trigger > Thresh')
            
            # Overlay: Mahalanobis Ellipse
            if intensity_sum > 1.0: # Only draw if valid signal
                ell = patches.Ellipse(
                    (com_x, com_y), 
                    width=r_major, height=r_minor, angle=theta_deg,
                    fill=False, edgecolor='magenta', linewidth=2.5, linestyle='-',
                    label=r'{s_sigma.value}$\sigma$ Boundary'
                )
                ax.add_patch(ell)
                ax.scatter([com_x], [com_y], c='white', edgecolors='black', s=80, label='CoM')
            
            ax.set_title(f"Frame {idx} | PE View", fontweight='bold')
            
            # Legend Panel
            ax_leg = fig.add_subplot(gs[1])
            ax_leg.axis('off')
            h, l = ax.get_legend_handles_labels()
            ax_leg.legend(h, l, loc='upper left', frameon=True)
            
            stats = (
                f"CONFIG\n"
                f"------\n"
                f"Sim Baseline: {s_base.value}\n"
                f"Sim Gain:     {s_gain.value}\n"
                f"PE Thresh:    {s_thresh.value}\n\n"
                f"ELLIPSE STATS\n"
                f"-------------\n"
                f"Center: ({com_x:.1f}, {com_y:.1f})\n"
                f"Spread: {r_major/2:.1f} x {r_minor/2:.1f} pix\n"
                f"Angle:  {theta_deg:.1f}°\n"
                f"Pixels Inside: {np.sum(mask)}"
            )
            props = dict(boxstyle='round', facecolor='whitesmoke', alpha=0.5)
            ax_leg.text(0.05, 0.3, stats, transform=ax_leg.transAxes, 
                        fontfamily='monospace', bbox=props, verticalalignment='top')
            
            plt.show()

    # Wiring
    drop_stream.observe(lambda c: init_stream(c['new']) if c['name']=='value' else None)
    for s in [s_base, s_gain, s_thresh, s_sigma, s_frame]:
        s.observe(update_view, names='value')
    
    def nav(b):
        step = 1 if b.description == "Next >" else -1
        s_frame.value = min(max(0, s_frame.value + step), s_frame.max)
    
    btn_next.on_click(nav)
    btn_prev.on_click(nav)
    
    # Layout
    ui_calib = widgets.HBox([s_base, s_gain])
    ui_algo = widgets.HBox([s_thresh, s_sigma])
    ui_nav = widgets.HBox([btn_prev, s_frame, btn_next])
    
    display(widgets.VBox([drop_stream, ui_calib, ui_algo, ui_nav, out]))
    
    if valid_prods: init_stream(valid_prods[0])

# Usage
# create_mahalanobis_visualizer(run)
```
