```python
from IPython.display import display, clear_output
import matplotlib.pyplot as plt
import numpy as np
import time
```

```python
def image_stream():
    for _ in range(10):
        yield np.random.rand(100, 100, 3)

for img in image_stream():
    plt.imshow(img)
    clear_output(wait=True)
    display(plt.gcf())
    time.sleep(0.5)
```
